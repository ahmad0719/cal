### **STRINGS**

```python
type(my_str)
```

```python
len(my_str)
```

#### **String Indexing/Slicing**

-indexes start from 0

-string[start:stop:step]

-starting index is included in the slice but stopping index is not

-starting index needs to be bigger than stopping index and step needs to be negative if you are slicing in reverse

-a step of 2 means skip every other element in the slice. a step of 3 means skip two elements

```python
my_str = "abcdefghijklmnopqrstuvwxyz"
```

```python
my_str[4]
```

```python
my_str[10]
```

```python
my_str[-2]
```

```python
my_str = "abcdefghijklmnopqrstuvwxyz"


print(my_str[3])
print(my_str[2:6])
print(my_str[:6])
print(my_str[6:])
print(my_str[0:11:1])
print(my_str[0:11:2])
print(my_str[0:11:3])
print(my_str[23:10:-1])
print(my_str[23:10:-2])
print(my_str[::-1])
```

```python
print(my_str[-2:-8:-2])
```

### **TUPLES**

-tuples are immutable

-defined by round brackets ()

-indexing of tuples works the same way as lists and strings

```python
my_tup = ("apple", "orange", "banana", "kiwi", "watermelon")
```

```python
type(my_tup)
```

```python
len(my_tup)
```

#### **Tuple methods**

https://www.w3schools.com/python/python_ref_tuple.asp

```python
my_tup.count("banana")
```

```python
my_tup.index("banana")
```

```python
# i want to insert melon at the 3rd index in my tuple
# single element tuple needs a comma at the end

my_tup = my_tup[:3] + ("melon",) + my_tup[3:]
```

```python
my_tup
```

### Importance of the copy method for mutable structures

Without copy()

```python
a = [1,2,3,4]

b = a

print(a)
print(b)
```

```python
a.append(5)
```

```python
print(a)
print(b)
```

With copy

```python
a = [1,2,3,4]

b = a.copy()

print(a)
print(b)
```

```python
a.append(5)
```

```python
print(a)
print(b)
```
